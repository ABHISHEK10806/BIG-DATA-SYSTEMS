#### 4. read the dataset

```python
df=spark.read.csv('/content/drive/MyDrive/SalesData.csv', header=True, inferSchema=True)

print("DataFrame loaded. Displaying the first 5 rows:")
df.show(5)
```

#### 5. Check the datatype of the dataframe

```python
print("DataFrame schema:")
df.printSchema()
```

#### 6. check the summary

```python
print("DataFrame summary statistics:")
df.describe().show()
```

#### 7. select column

```python
# Select a single column
print("Selecting '_c11' (Total Revenue) column:")
df.select('_c11').show(5)

# Select multiple columns
print("Selecting '_c11' (Total Revenue) and '_c12' (Total Expense) columns:")
df.select('_c11', '_c12').show(5)
```

#### 8. Adding column

```python
from pyspark.sql.functions import col, lit
from pyspark.sql.types import DoubleType

# Adding a new column 'revenue_per_unit' using existing numeric columns
# Casting 'Total Revenue' and 'Units Sold' to DoubleType for calculation
df_with_new_col = df.withColumn(
    'revenue_per_unit',
    (col('Total Revenue').cast(DoubleType()) / col('Units Sold').cast(DoubleType()))
)

print("DataFrame with new 'revenue_per_unit' column:")
df_with_new_col.select('Units Sold', 'Total Revenue', 'revenue_per_unit').show(5)

# Adding a constant column
df_with_constant = df.withColumn('source', lit('Sales Data'))
print("DataFrame with new 'source' column:")
df_with_constant.select('source').show(5)
```

#### 9. Dropping column


```python
# Drop a single column
df_after_dropping_one = df_with_new_col.drop('revenue_per_unit')
print("DataFrame after dropping 'revenue_per_unit' column:")
df_after_dropping_one.printSchema()
df_after_dropping_one.show(5)

# Drop multiple columns
df_after_dropping_multiple = df_with_constant.drop('source', 'Order Priority')
print("\nDataFrame after dropping 'source' and 'Order Priority' columns:")
df_after_dropping_multiple.printSchema()
df_after_dropping_multiple.show(5)
```

#### 10.Renaming column

```python
# Renaming 'Order Priority' to 'order_priority_level'
df_renamed = df.withColumnRenamed('Order Priority', 'order_priority_level')

print("DataFrame after renaming 'Order Priority' to 'order_priority_level'. Schema before rename:")
df.printSchema()
print("\nSchema after rename:")
df_renamed.printSchema()

# To see the effect:
df_renamed.select('order_priority_level').show(5)
```
