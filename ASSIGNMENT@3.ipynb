## 1. Load Data (START SPARK SESSION)

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("EmployeeOperations").getOrCreate()

# Example dataset
data = [
    ("Aman", 28, 40000, "Delhi"),
    ("Ravi", 35, 60000, "Mumbai"),
    ("Neha", 32, 55000, "Delhi"),
    ("Sita", 29, 70000, "Bangalore"),
    ("Karan", 40, 80000, "Delhi")
]

columns = ["name", "age", "salary", "city"]

df = spark.createDataFrame(data, columns)

df.show()
```

##2. Filter (Salary > 50000)

```python
df.filter(df.salary > 50000).show()
```

##3. Filter (Age > 30)

```python
df.filter(df.age > 30).show()
```

##4. Count

```python
df.count()
```

##5. Filter (City == Delhi)

```python
df.filter(df.city == "Delhi").show()
```

##6. Sample Dataset

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import count, sum, avg, max

spark = SparkSession.builder.appName("GroupByOperations").getOrCreate()

data = [
    ("Aman", 28, 40000, "IT", "Male", "Delhi"),
    ("Ravi", 35, 60000, "HR", "Male", "Mumbai"),
    ("Neha", 32, 55000, "IT", "Female", "Delhi"),
    ("Sita", 29, 70000, "Finance", "Female", "Bangalore"),
    ("Karan", 40, 80000, "IT", "Male", "Delhi"),
    ("Pooja", 27, 45000, "HR", "Female", "Mumbai")
]

columns = ["name", "age", "salary", "department", "gender", "city"]

df = spark.createDataFrame(data, columns)

df.show()
```

 ## 7. GroupBy (Department & Employees_Count)

```python
df.groupBy("department").agg(count("*").alias("employee_count")).show()
```

## 8. Total_salary per Department

```python
df.groupBy("department").agg(sum("salary").alias("total_salary")).show()
```

## 9. Avg_salary per Department

```python
df.groupBy("department").agg(avg("salary").alias("avg_salary")).show()
```

 ## 10. GroupBy (City & find max_salary)

```python
df.groupBy("city").agg(max("salary").alias("max_salary")).show()
```
